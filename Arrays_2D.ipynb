* 2D Array Input

```python
import numpy as np

rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

arr = []

for i in range(rows):
    row = list(map(int, input(f"Enter elements of row {i + 1}: ").split()))
    arr.append(row)

arr = np.array(arr)

print("2D Array:")
print(arr)

print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Total Elements:", arr.size)
```

* 2D Array Traversal

```python
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

arr = []

for i in range(rows):
    row = list(map(int, input(f"Enter elements of row {i + 1}: ").split()))
    arr.append(row)

arr = np.array(arr)

for i in range(rows):
    for j in range(cols):
        print(arr[i][j] , end = " ")
    print()
```

* Linear search

```python
arr
target = int(input("Enter Target Element: "))

for i in range(rows):
    for j in range(cols):
        if arr[i][j] == target:
            print("Element Found at Index ",(i, j))
            break
    else:
        continue
    break
```
