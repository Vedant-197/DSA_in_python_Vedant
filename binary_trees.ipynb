* Height , Node Count , Sum of all Nodes

```python
def create_node(data):
    return {
        "data" : data,
        "left" : None,
        "right": None
    }

def build_tree(arr):
    global idx
    idx += 1

    if arr[idx] == -1:
        return None
    root = create_node(arr[idx])
    root["left"] = build_tree(arr)
    root["right"] = build_tree(arr)
    return root

arr = list(map(int, input("Enter Preorder values: ").split()))

idx = -1
root = build_tree(arr)

def height_bt(root):           # Height
    if root is None:
        return 0
    left_height = height_bt(root["left"])
    right_height = height_bt(root["right"])

    return (max(left_height, right_height) + 1)

def count_nodes(root):      # count nodes
    if root is None:
        return 0
    left_count = count_nodes(root["left"])
    right_count = count_nodes(root["right"])

    return (left_count + right_count + 1)

def sum_of_nodes(root):                  # sum of nodes
    if root is None:
        return 0

    left_sum = sum_of_nodes(root["left"])
    right_sum = sum_of_nodes(root["right"])

    return (left_sum + right_sum + root["data"])

print(f"Binary Tree Inputs: ")
print(arr)
print(" ")
print("Height of the tree is:", height_bt(root))
print("Total number of nodes is:", count_nodes(root))
print("Sum of all nodes is:", sum_of_nodes(root))
```

* Top View of Binary Tree

```python
from collections import deque

def create_node(data):
    return {
        "data": data,
        "left": None,
        "right": None
    }


def build_tree(arr):
    global idx
    idx += 1

    if arr[idx] == -1:
        return None

    root = create_node(arr[idx])
    root["left"] = build_tree(arr)
    root["right"] = build_tree(arr)

    return root


def top_view(root):
    if root == None:
        return

    m = {}
    q = deque()

    q.append((root, 0))

    while q:
        curr, currHD = q.popleft()

        if currHD not in m:
            m[currHD] = curr["data"]

        if curr["left"] != None:
            q.append((curr["left"], currHD - 1))

        if curr["right"] != None:
            q.append((curr["right"], currHD + 1))

    for i in sorted(m):
        print(m[i], end=" ")


arr = list(map(int, input("Enter Preorder values: ").split()))
idx = -1

root = build_tree(arr)
print("Binary Tree Preorder Form Inputs : ")
print(arr)
print("\nBinary tree: ")
print_tree(root,space=0)

print("\nTop View of the Binary Tree:")
top_view(root)
```

* K^th Order of Binary Tree 

```python
def k_order(root,k):
    if root == None:
        return
    if k == 1:
        print(root["data"], end = " ")
        return
    k_order(root["left"], k-1)
    k_order(root["right"], k-1)

arr = list(map(int, input("Enter Preorder values: ").split()))
idx = -1

root = build_tree(arr)
print("Binary Tree Preorder Form Inputs : ")
print(arr)
print("\nBinary tree: ")
print_tree(root,space=0)

k = int(input("Enter K: "))
print("\nk: ",k)
print("Nodes at Kth Level:")
k_order(root, k)
```

* Lowest Common Ancestor

```python

def lowest_common_ancestor(root, p, q):
    if root == None:
        return None

    if root == p or root == q:
        return root

    leftLCA = lowest_common_ancestor(root["left"], p, q)
    rightLCA = lowest_common_ancestor(root["right"], p, q)

    if leftLCA != None and rightLCA != None:
        return root

    if leftLCA != None:
        return leftLCA

    return rightLCA
```

* Transformation of Tree to Sum Tree

```python

def transform_to_sum_tree(root):
    if root == None:
        return 0

    leftSum = transform_to_sum_tree(root["left"])
    rightSum = transform_to_sum_tree(root["right"])

    root["data"] += leftSum + rightSum

    return root["data"]


def print_preorder(root):
    if root == None:
        return

    print(root["data"], end=" ")
    print_preorder(root["left"])
    print_preorder(root["right"])


arr = list(map(int, input("Enter Preorder values: ").split()))
idx = -1

root = build_tree(arr)

print("Preorder Traversal BEFORE transformation:")
print_preorder(root)

transform_to_sum_tree(root)

print("\nPreorder Traversal AFTER transformation:")
print_preorder(root)
```
