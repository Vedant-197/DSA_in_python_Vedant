* Print Numbers from n to 1

```python
n = int(input("Enter the Number: "))     # n = 5

def printnum(n):
    if n == 1:
        print(n)
        return
    else:
        print(n,end = " ")
        printnum(n-1)

printnum(n)
```

* Print Number from 1 to n

```python
n = int(input("Enter the Number: "))

def printnumbers(n):
    if n == 1:
        print(1, end=" ")
    else:
        printnumbers(n - 1)
        print(n, end=" ")

print(f"Numbers from 1 to {n}: ")
printnumbers(n)
```

* N Factorial

```python
n = int(input("Enter the Number: "))

def factorial(n):
    if n == 0:
        return(1)
    else:
        return n * factorial(n - 1)

print(f"Factorial of {n} is {factorial(n)}")
```

* Sum of N Numbers

```python
n = int(input("Enter the Number: "))

def sum(n):
    if n == 1:
        return 1
    else:
        return n + sum(n - 1)

print(f"The Sum of {n} Numbers is {sum(n)}")
```

* Fibonacci Series

```python
n = int(input("Enter the Term: "))

def fibo(n):
    if ((n == 0) or (n == 1)):
        return n
    else:
        return fibo(n - 1) + fibo(n - 2)

print(f"Fibonacci Number of Term {n} is {fibo(n)}")

#Series print
print(f"Fibonacci series: ")
for i in range(n + 1): 
    print(fibo(i), end = " ")
```

* Check whether Array is Sorted or not

```python
import numpy as np
arr = np.array(list(map(int,input("Enter Elements of an Array Seperated by space: ").split())))
n = len(arr)
def is_array_sorted(arr,n):
    if (n == 0) or (n == 1):
        return True
    else:
        return (arr[n-1] >= arr[n-2]) and is_array_sorted(arr,n-1)

print("Orignal Array: ")
print(arr, end = " ")
print(" ")
if is_array_sorted(arr,n)  == True:
    print("Array is Sorted")
else: 
    print("Array is not Sorted")
```

*   Binary search Algorithm using Recursion

```python
arr = np.array(list(map(int,input("Enter Elements in Sorted Order Seperated by space: ").split())))
target = int(input("Enter Target: "))
n = len(arr)
st = 0
end = n - 1

def bs(arr, st , end , target):
    mid = st + (end - st)//2
    if (arr[mid] == target):
        return mid
    elif (arr[mid] <= target):
        return bs(arr,mid + 1, end, target)
    else:
        return bs(arr,st, mid - 1,target)

print("Orignal Array: ")
print(arr, end = " ")
print(" ")
print(f"Target Element {target} is Found at index {bs(arr, st, end, target)}")
```

* Printing all Subsets of an Array

```python
def printSubsets(arr, ans, i):
    if i == len(arr):
        print(ans)
        return

    ans.append(arr[i])
    printSubsets(arr, ans, i + 1)
    ans.pop()

    printSubsets(arr, ans, i + 1)


arr = list(map(int, input("Enter array elements: ").split()))
ans = []
printSubsets(arr, ans, 0)
```
