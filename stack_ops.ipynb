Stsck Input by Split and append

```python
# using Split

s = list(map(int,input("Enter Elements of Stack: ").split()))

print(f"Stack by inputs: {s}")
print("Elements of Stack from Top to Bottom: ")

while len(s) > 0:
    print(s[-1],end = " ")
    s.pop()
    
print()
```

```python
# Using Append

s = []
n = int(input("Enter no of Elements: "))
for i in range(n):
    x = int(input("Enter Element of Stack: "))
    s.append(x)

print(f"Stack by inputs: {s}")
print("Stack From Top to Bottom: ")    
while len(s) > 0:
    print(s[-1], end = " ")
    s.pop()
print()

```

Pop 

```python
s = list(map(int, input("Enter Elements: ").split()))
print("Stack: ", s)
if len(s) > 0 :
    x = s.pop()
    print("Popped Element: ",x )
    print("Stack After Pop : ",s)
else:
    print("Stack is Empty")
```

Display

```python
s =  list(map(int, input("Enter Elements: ").split()))
print("Stack: ",s)
print("Top to Bottom: ")

for i in range((len(s)-1),-1,-1):
    print(s[i],end = " ")
print()
```

Search

```python
s = list(map(int, input("Enter Elements: ").split()))
x = int(input("Enter Element to Search: "))
print("Stack: ", s)
print(f"You Entered {x}")
if x in s:
    print(f"Element Found at {s.index(x)}")
else:
    print("Element not Found")
```

Swap First and last Index

```python
s = list(map(int, input("Enter Elements: ").split()))
print("Stack: ",s)
if len(s) > 1:
    s[0],s[-1] = s[-1],s[0]
print("AFter Swapping: ", s)
```

Remove Duplicates

```python
s = list(map(int, input("Enter Elements: ").split()))
print("Stack: ",s)
t = []
for i in s:
    if i  not in t:
        t.append(i)
print("Stack Without Duplicates: ",t)
```

Seperate Odd and Even Function

```python
s = list(map(int, input("Enter Elements: ").split()))
print("Stack: ",s)
even = []
odd = []
for i in s:
    if (i % 2) == 0:
        even.append(i)
    else:
        odd.append(i)
print("Even Stack : ", even)
print("Odd Stack : ", odd) 
```
